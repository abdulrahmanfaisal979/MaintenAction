```python
import serial
import csv
import time

PORT = "COM3"
BAUD_RATE = 115200

ACTION = "test"
TRIAL = 3

filename = rf"C:\Users\DELL\Desktop\{ACTION}_{TRIAL:02d}.csv"

ser = serial.Serial(PORT, BAUD_RATE, timeout=1)
time.sleep(2)

print("Recording...")
print("Do: 2 sec still → action → 2 sec still")
print("Press Ctrl+C when finished")

with open(filename, "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow([
        "time_ms",
        "ax",
        "ay",
        "az",
        "gx",
        "gy",
        "gz",
        "label"
    ])

    try:
        while True:
            line = ser.readline().decode("utf-8", errors="ignore").strip()
            values = line.split(",")

            if len(values) == 7:
                try:
                    numbers = [int(x) for x in values]
                    writer.writerow(numbers + [ACTION])
                except ValueError:
                    pass

    except KeyboardInterrupt:
        print("\nRecording stopped.")

ser.close()

print("Saved as:", filename)
```
